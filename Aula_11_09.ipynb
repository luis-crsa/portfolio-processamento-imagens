```python
img_rgb = cv2.imread('mentalista.png', cv2.IMREAD_COLOR_RGB)
plt.imshow(img_rgb)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
def calHist(img):
  hist_arr = np.zeros(256, dtype=int)
  for px in img:
    hist_arr[px] += 1
  plt.bar(range(256), hist_arr)
```

```python
calHist(img_gray)
```
